## 1. Consulta l’API pública JSONPlaceholder utilitzant el mètode GET per obtenir:

- Llista de publicacions (/posts)
- Llista d’usuaris (/users)
- Llista de tasques (/todos)

```python
import pandas as pd
import requests
import json
```

```python
response_posts = requests.get("https://jsonplaceholder.typicode.com/posts")
posts = response_posts.json()
print(posts)
```

```python
response_users = requests.get("https://jsonplaceholder.typicode.com/users")
users = response_users.json()
print(users)
```

```python
response_todos = requests.get("https://jsonplaceholder.typicode.com/todos")
todos = response_todos.json()
print(todos)
```

## 2. Mostra per pantalla:

- La quantitat total de cada recurs.
- El codi d'estat de cada petició.

```python
print(f"Quantitat de posts: {len(posts)} \nQuantitat de usuaris: {len(users)} \nQuantitat de todos: {len(todos)}")
```

```python
print(f"Codi estat posts: {response_posts.status_code} \nCodi estat usuaris: {response_users.status_code} \nCodi estat todos: {response_todos.status_code}")
```

## 3. Fes una petició a una publicació inexistent per obtenir un error 404 i mostra el codi d'estat rebut.

```python
response_failed = requests.get("https://jsonplaceholder.typicode.com/company")
print(f"Codi estat: {response_failed.status_code}")
```

## 4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

- Mostra la resposta JSON.
- Mostra el codi d'estat.


```python
response_add = requests.post("https://jsonplaceholder.typicode.com/posts", json = {'userId': 1, 'title': 'Nuevo post', 'body': 'Nuevo post'})
print(f"Resposta JSON: {response_add.json()} \nCodi estat: {response_add.status_code}")
```

## 5. Fes una petició PATCH per modificar parcialment una publicació existent.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
response_patch = requests.patch("https://jsonplaceholder.typicode.com/posts/1", json = {'title': 'Nuevo título'})
print(f"Resposta JSON: {response_patch.json()} \nCodi estat: {response_patch.status_code}")
```

## 6. Fes una petició DELETE sobre una publicació.

- Mostra la resposta JSON.
- Mostra el codi d'estat.

```python
response_delete = requests.delete("https://jsonplaceholder.typicode.com/posts/1")
print(f"Resposta JSON: {response_delete.json()} \nCodi estat: {response_delete.status_code}")
```
