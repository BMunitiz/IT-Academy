## 1. Calculadora de l'índex de massa corporal

Escriu una funció que calculi l'IMC ingressat per l'usuari/ària, és a dir, qui ho executi haurà d'ingressar aquestes dades. Pots obtenir més informació del seu càlcul en: 🔗 Índice de masa corporal IMC que es y como se calcula.
La funció ha de classificar el resultat en les seves respectives categories.
Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.

```python
def calc_imc():
    try:
        peso = float(input("intruduce tu peso en kg: "))
        altura = float(input("Introduce tu altura en metros: "))
        if (0.0 < peso < 350.0) & (0.0 < altura < 2.8): 
            imc = peso / altura**2
            if imc < 18.5:
                rango = "Peso bajo" 
            elif 18.5 <= imc <= 24.9:
                rango = "Peso normal"
            elif 25 <= imc <= 29.9: 
                rango = "Sobrepeso" 
            else:
                rango = "Obesidad"
            return print(f"IMC: {round(imc,2)} rango: {rango}"), imc, rango
        else:
            print("El peso debe estar entre 0 y 350, la altura debe estar entre 0 y 2.8")
        
    except ValueError:
        print("El peso y la altura deben ser números.")
   
```

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un print() no es pot guardar mai).

```python
def temp_conversor():
    scales = ["C", "F", "K", "R"]
    try:
        temp = float(input("Introduce la temperatura: ")) 
        unidad = str.upper(input("Introduce la unidad de temperatura: "))
        v = temp
        if unidad in scales:
            conversions = {"C": [ lambda v: (v * 9/5) + 32, # C a F 
                                  lambda v: v - 273.15, # C a K
                                  lambda v: (v + 273.15) * 9/5], # C a R
                           "F": [ lambda v: (v - 32) * 5/9, # F a C
                                  lambda v: (v + 459.67) * 5/9, # F a K
                                  lambda v: (v * 9/5) - 459.67], # F a R
                           "K": [ lambda v: v + 273.15, # K a C
                                  lambda v: (v * 9/5) - 459.67, # K a F
                                  lambda v: (v * 9/5)], # K a R
                           "R": [ lambda v: (v - 491.67) * 5/9, # R a C
                                  lambda v: (v * 9/5) - 459.67, # R a F
                                  lambda v: (v * 9/5) - 459.67] # R a K
                          } 
            func = conversions[unidad]
            lista = []
            for x in func:
                y = x(temp)
                lista.append(y)
            if unidad == "C":
                print(f"{temp}{unidad} equivalen a {round(lista[0],2)}F, {round(lista[1],2)}K y {round(lista[2],2)}R")
            elif unidad == "F":
                print(f"{temp}{unidad} equivalen a {round(lista[0],2)}C, {round(lista[1],2)}K y {round(lista[2],2)}R")
            elif unidad == "K":
                print(f"{temp}{unidad} equivalen a {round(lista[0],2)}C, {round(lista[1],2)}F y {round(lista[2],2)}R")
            else:
                print(f"{temp}{unidad} equivalen a {round(lista[0],2)}C, {round(lista[1],2)}F y {round(lista[2],2)}K")
                
            return lista
        else:
            print("Unidad de temperatura incorrecta")

    except ValueError: 
        print ("La temperatura debe ser un número") 
```

## 3. Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.

(EXTRA): Quina és la llargada mitjana de les paraules del text que has escrit? “Hola com va?” hauria de retornar (4+3+2) / 3 = 3

```python
import re
def contador_palabras(text):
    if text.strip(): 
        text = text.lower()
        text = re.sub(r'[^\w\s]', '', text)
        text = text.replace("\n", " ")
        length = text.split(" ")
        words = {}
        letter_counter = {}
        letter_sum = 0
        for word in length:
            letter_counter[word] = len(word)
            letter_sum += letter_counter[word]
            if word in words:
                words[word] += 1
            else:
                words[word] = 1
            
        avg_letters = letter_sum/len(length)
        #print(f"Repetición de palabras: {words}, media de letras por palabra: {avg_letters}")


    else:
        print("El texto no puede estar vacio")
    return words, avg_letters
```

## 4. Diccionari invers (amb possibilitat de duplicats)

Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors. Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.

```python
dict01 = {"a":1, "b":2, "c":3}
dict02 = {"x": "apple", "y": "banana", "z": "banana" }
def dict_invers(dictionary):
    lista=[]
    correct_dict = {}
    for key, value in dictionary.items():
        if value in correct_dict:
            if isinstance(correct_dict[value], list):
                correct_dict[value].append(key)
            else:
                correct_dict[value]= [correct_dict[value],key]
            lista.append(value)
        else:
            correct_dict[value] = key
    if lista:
        print(f"Se han encontrado valores repetidos, los valores repetidos son: {lista}")
    return correct_dict
```

```python
import string
def word_order(text):
    new_dict = {}
    ordered_dict = {}
    alphabet = list(string.ascii_lowercase)
    word_count, _ = contador_palabras(text.read())
    word_ordered = dict(sorted(word_count.items()))
    for key, value in word_ordered.items():
        for letter in alphabet:
            if key != " ":
                if key.startswith(letter):
                    ordered_dict.setdefault(letter, {}).setdefault(key , value)
        
    return ordered_dict
```

## 2. Conversió de tipus de dades.

El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir.

```python
def float_no_float_lists(lista):
    float_lista = []
    no_float_lista = []
    for element in lista:
        if type(element) is list:
            for x in element:
                try:
                    float_lista.append(float(x))
                except:
                    no_float_lista.append(x)
                
        elif type(element) is tuple:
            for y in element:
                try:
                    float_lista.append(float(y))
                except:
                    no_float_lista.append(y)       
        else:
            try:
                float_lista.append(float(element))
            except:
                no_float_lista.append(element)
    return float_lista, no_float_lista

        
        
```

Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy

En aquest punt, el client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

longitud (int): Longitud de la contrasenya
majuscules (bool = True): Si hi ha d’aparèixer majúscules
minuscules (bool = True): Si hi ha d’aparèixer minúscules
numeros (bool = True): Si hi ha d’aparèixer números
signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)

```python
from numpy import random
import string
import os
import clipboard


def password(length: int, upper: bool, lower: bool, number: bool, signs: bool):
    password = ""
    rand_num = random.randint(1, length)
    lista = ""
    if upper == True:
        lista += string.ascii_uppercase
    if lower == True:
        lista += string.ascii_lowercase
    if number == True:
        lista += string.digits
    if signs == True:
        lista += string.punctuation
    #char = random.choice(lista)
    for x in range(length):
        random_index = int(random.random() * len(lista))
        password += lista[random_index]
    clipboard.copy(password)
    return password
    
```

## 2. Processament de dades simple

Una companya de feina ens ha demanat un favor, aprofitant que sap que estem aprenent a programar. Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.

```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

data = pd.read_csv("historic partits.txt", sep= '\t')
data.columns = ["casa", "resultat", "fora"]
data[["gols casa", "gols fora"]] = data.resultat.str.split("-", expand=True)
data[["gols casa", "gols fora"]] = data[["gols casa", "gols fora"]].apply(pd.to_numeric)
column_names = ["Partits guanyats", "Partits empatats", "Partits perduts", "Gols", "Gols encaixats", "Punts"]
teams = data.casa.unique()
final_data = pd.DataFrame(index = teams, columns = column_names)
for col in column_names:
    final_data[col] = 0

for index, item in data.iterrows():
    if item["gols casa"] > item["gols fora"]:
        final_data['Gols'][[item['casa']]] += item['gols casa']
        final_data['Gols encaixats'][[item['casa']]] += item['gols fora']
        final_data['Gols encaixats'][[item['fora']]] += item['gols casa']
        final_data['Gols'][[item['fora']]] += item['gols fora']
        final_data['Punts'][[item['casa']]] += 3
        final_data['Partits guanyats'][[item['casa']]] +=1
        final_data['Partits perduts'][[item['fora']]] +=1
        
    elif item["gols casa"] < item["gols fora"]:
        final_data['Gols'][[item['casa']]] += item['gols casa']
        final_data['Gols encaixats'][[item['casa']]] += item['gols fora']
        final_data['Gols encaixats'][[item['fora']]] += item['gols casa']
        final_data['Gols'][[item['fora']]] += item['gols fora']
        final_data['Punts'][[item['fora']]] +=3   
        final_data['Partits guanyats'][[item['fora']]] +=1
        final_data['Partits perduts'][[item['casa']]] +=1
    
    else:
        final_data['Gols'][[item['casa']]] += item['gols casa']
        final_data['Gols encaixats'][[item['casa']]] += item['gols fora']
        final_data['Gols encaixats'][[item['fora']]] += item['gols casa']
        final_data['Gols'][[item['fora']]] += item['gols fora']
        final_data['Punts'][[item['casa']]] +=1
        final_data['Punts'][[item['fora']]] +=1
        final_data['Partits empatats'][[item['casa']]] +=1
        final_data['Partits empatats'][[item['fora']]] +=1
        
final_data = final_data.sort_values(by=['Punts'],ascending = False)
top = final_data.idxmax()

print(f'Gols per equip: \n{final_data['Gols']}')
print(f'Equip més golejador: {top['Gols']}\nEquip més golejat: {top['Gols encaixats']}')
print(f'Classificació global: \n{final_data['Punts']}')
print (final_data)
```
